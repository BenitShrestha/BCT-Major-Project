```python
from rouge_score import rouge_scorer
from rouge_score.tokenizers import Tokenizer
import re

# =========================
# Tokenizers
# =========================

class NepaliTokenizer1(Tokenizer):
    """Keeps all word characters, language-agnostic"""
    def tokenize(self, text):
        text = text.replace("।", "")
        text = re.sub(r"[^\w\s]", "", text)
        return text.split()

class NepaliTokenizer2(Tokenizer):
    """Strict Nepali: normalize digits, keep only Devanagari"""
    def tokenize(self, text):
        text = text.replace("।", "")
        text = re.sub(r"[०-९]+", "NUM", text)  # Normalize Nepali digits
        text = re.sub(r"[^\u0900-\u097F\s]", "", text)  # Keep only Devanagari + space
        return text.split()

# =========================
# ROUGE Scorers
# =========================

tokenizer1 = NepaliTokenizer1()
tokenizer2 = NepaliTokenizer2()

scorer1 = rouge_scorer.RougeScorer(
    ['rouge1', 'rouge2', 'rougeL'],
    tokenizer=tokenizer1,
    use_stemmer=False
)

scorer2 = rouge_scorer.RougeScorer(
    ['rouge1', 'rouge2', 'rougeL'],
    tokenizer=tokenizer2,
    use_stemmer=False
)

# =========================
# Sample summaries
# =========================

reference_summary = "भारत सरकारले प्रमुख विपक्षी दल कंग्रेसका अध्यक्ष राहुल गान्धीलाई उनीसँग विदेशी नागरिकता भए नभएको स्पष्ट पार्न भनेको छ।"
generated_summary = "भारतमा कोरोनाभाइरस सङ्क्रमितको सङ्ख्या गर्दै-१९ गान्धीले पहिरो विद्यमानबाटरको छ। यो हुनेबाट तथा त्यसको बेला, गरेको छ। यो"

# =========================
# Display tokenized/transformed texts
# =========================

tokens_ref1 = tokenizer1.tokenize(reference_summary)
tokens_gen1 = tokenizer1.tokenize(generated_summary)

tokens_ref2 = tokenizer2.tokenize(reference_summary)
tokens_gen2 = tokenizer2.tokenize(generated_summary)

print("=== Tokenizer 1 (Permissive) Transformed Texts ===")
print("Reference:", " ".join(tokens_ref1))
print("Generated:", " ".join(tokens_gen1))

print("\n=== Tokenizer 2 (Strict Nepali) Transformed Texts ===")
print("Reference:", " ".join(tokens_ref2))
print("Generated:", " ".join(tokens_gen2))

# =========================
# Score computation
# =========================

scores1 = scorer1.score(reference_summary, generated_summary)
scores2 = scorer2.score(reference_summary, generated_summary)

# =========================
# Print ROUGE results
# =========================

print("\n=== Tokenizer 1 (Permissive) ROUGE Scores ===")
print("ROUGE-1:", scores1['rouge1'])
print("ROUGE-2:", scores1['rouge2'])
print("ROUGE-L:", scores1['rougeL'])

print("\n=== Tokenizer 2 (Strict Nepali) ROUGE Scores ===")
print("ROUGE-1:", scores2['rouge1'])
print("ROUGE-2:", scores2['rouge2'])
print("ROUGE-L:", scores2['rougeL'])
```
